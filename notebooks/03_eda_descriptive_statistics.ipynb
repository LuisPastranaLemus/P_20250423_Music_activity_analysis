## __5. Insights__

### 🎯 Statistical Findings

- **High Variability Across Day, Hour, Genre, Artist, Tracks**:  
  Coefficient of Variation (CV) highlights notable variability in activity patterns. Making mean be misleading, recommending to use Median as
  reference central value.

- **Segmentation**:  
  Day and hour show a clear segmentation pattern in music activity, indicating that the resulting groups should be further analyzed.

- **Correlation**:  
  The strong correlation between Genre, Track, and Artist reflects genre specialization: most artists concentrate their catalog within a single genre, which can simplify user segmentation and recommendation models

### 🎯 Preliminary Business Findings

- **Most Frequent Played Track, Artist, Genre**:   
  Track: 'brand'   
  Artist: 'kartvelli'   
  Genre: 'pop'   

- **Location Music Activity**:
  Springfield stands out with higher levels of music activity than Spriengfield, suggesting stronger audience engagement in that area.

- **Date And Time Music Activity**:
  Music consumption peaks in the Afternoon and reaches its highest levels on Fridays, suggesting strong end-of-week engagement.

- **Trends in music genres across locations**:
  While Springfield and Shelbyville show different levels of music activity, both exhibit similar preferences in genres, mainly Pop, Dance, Electronic, and Rock.

  ***Note**:
  These findings are preliminary. Accounting for outliers and performing additional data segmentation will help achieve clearer and more reliable insights.
